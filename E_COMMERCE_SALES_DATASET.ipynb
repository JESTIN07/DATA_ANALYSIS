# 2. Which region has the highest number of orders?

```python
sns.countplot(sales,x='region')
print("THE REGION WITH THE HIGHEST NUMBER OF ORDERS IS",sales['region'].value_counts().idxmax(),"WITH",sales['region'].value_counts().max(),"ORDERS")
```

# CODE DEFENITION 
sales['region'].value_counts().idxmax() = FROM THE DATASET SALES, COUNT ALL THE VALUE AND RETURN THE REGION WHICH HAS THE MAX VALUE
sales['region'].value_counts().max() = FROM THE DATASET SALES, COUNT ALL THE VALUE AND RETURN THE MAX VALUE

# 3). Which product categories have the highest number of orders in each region?

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.countplot(data=sales, x='region', hue='product_category')
plt.legend(title='Product Category',fontsize=5)
```

# IN ALL FOUR REGIONS ELECTRONIC PRODUCT IS THE MOST ORDERED PRODUCT AND BEAUTY PRODUCT IS THE LESS ORDERED PRODUCT. CLOTHING PRODUCT AS THE SECOND AND HOME PRODUCT AS THE THIRD

# 4) How does revenue vary across product categories in different regions?

```python
sns.barplot(data=sales,x='region',y='revenue',hue='product_category')
```

X AXIS = REGION , Y AXIS = REVENUE , FOUR BARS = REPRESENT EACH PRODUCT
-> IN THE SOUTHERN REGION BEAUTY PRODUCT IS SOLDED FOR HIGH PRICE AND HOME PRODUCT IS SOLDED FOR LOW PRICE
-> IN THE NORTHERN REGION HOME PRODUCT IS SOLDED FOR HIGHER PRICE AND BEAUTY PRODUCT IS SOLDED FOR LOW PRICE

# How do sales/revenue change over time?

```python
import pandas as pd 
sales['order_date']=pd.to_datetime(sales['order_date'])
sns.lineplot(data=sales,x='order_date',y='revenue')

plt.figure(figsize=(12,5))
plt.title('YEARLY BASED ANALYSIS')
plt.xlabel("ORDER YEAR")
plt.ylabel("REVENUE")
```
